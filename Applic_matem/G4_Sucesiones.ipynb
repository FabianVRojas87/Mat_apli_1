# Problema 4

Implementa un código que permita guardar en una lista la sucesión de los primeros $n$ números pares, con $n \geq 1$. Luego, determina:

1. ¿Cuál es el valor del décimo término?
2. Calcula, utilizando un ciclo *for*, la suma de los primeros $100$ números pares.
3. ¿En qué posición se encuentra el número $58$ dentro de la sucesión?

```python
#1
def a(n):
    return 2*n
print (f'El  termino de posicion 10 es {a(10)}')

#2
suma=0
for i in range(1,100,1):
    suma = suma + a(i)
print(f'La suma de los primeros 100 pares es {suma}')

#3
#En que posicion esta el termino 58:
#Resolviendo una ecuacion : a(n)=58

import sympy as sp

n=sp.Symbol('n') #'variable de la ecuación
ecuacion = sp.Eq(a(n),58) #ecuacion
solucion = sp.solve(ecuacion,n)
print(solucion)

#otra solucion.....
contador = 0
while a(contador) < 58:
    contador += 1
print(contador)

```

```python
#formula para una sucesion aritmetica
#  a_n = a_1 + (n-1) *d

#         a_n=120000+(n-1)*2000

# Es una sucesion aritmetica porque aumenta de forma constante en 2000 Ademas, la diferencia entre los terminos  consecutivos es: d= 2000
#recordar que la formula de una S.A (sucecion aritmetica) es : #  a_n = a_1 + (n-1) *d

def a(n):
    return 12000+ (n-1)*2000

#1 
for n in range (1,13):
    print (f'El mes {n}  {a(n)}')

#2 : febrero de 2do año es la posicion 14:

print(f'En febrero del 2do año depositó {a(14)} pesos')

#3.1 : Debemos sumar desde el mes 1 al 24 
# primera forma

suma= 0

for i in range (1,25):
    suma = suma + a (i)
print (f'El total ahorrado al finalizar en dos años es de {suma} pesos')

#3.2 :usando listas y sumando los elementos de la lista
# Otra forma

lista=[]

for i in range (1,25):
    lista.append(a(i))
print(lista)
suma_lista = sum (lista)
print(suma_lista)

# 4  cuanto ahorro en total duran el primer año
# otra forma de sumar en la lista , recorre la lista y se posiciona el cual se debe sumar}

print (sum(lista[:2])) 
print (sum(lista[:12])) # Elk total  durante el primer año
print (sum(lista[0:2]))
print (sum(lista[2:4]))   # el total entre tercer y cuarto mes






```
